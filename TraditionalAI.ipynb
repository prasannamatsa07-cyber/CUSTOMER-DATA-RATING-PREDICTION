```python
# Traditional AI: Rule-Based E-Commerce Recommendation System

print("=== E-Commerce Product Recommendation ===")

budget = int(input("Enter your budget (₹): "))
category = input("Enter product category (laptop/phone/shoes): ").lower()

# Rule 1: Laptop
if category == "laptop" and budget >= 50000:
    print("Recommendation: Buy a Premium Laptop")

# Rule 2: Laptop
elif category == "laptop" and budget < 50000:
    print("Recommendation: Buy a Budget Laptop")

# Rule 3: Phone
elif category == "phone" and budget >= 30000:
    print("Recommendation: Buy a Premium Smartphone")

# Rule 4: Phone
elif category == "phone" and budget < 30000:
    print("Recommendation: Buy a Budget Smartphone")

# Rule 5: Shoes
elif category == "shoes" and budget >= 3000:
    print("Recommendation: Buy Branded Sports Shoes")

elif category == "shoes" and budget < 3000:
    print("Recommendation: Buy Affordable Sports Shoes")

else:
    print("Sorry, no recommendation available.")
```
